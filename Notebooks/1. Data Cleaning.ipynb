```python
# Fill missing values in the "Phases" column for non-interventional studies to "Not Applicable"
mask = df["Study Type"] != "Interventional"

df.loc[mask, "Phases"] = (
    df.loc[mask, "Phases"]
      .fillna("Not Applicable")
)
```

```python
print(df.groupby('Study Type')['Phases'].apply(lambda x: x.isnull().sum()))
```

```python
date_columns = [
    "Start Date",
    "Primary Completion Date",
    "Completion Date",
    "First Posted",
    "Results First Posted",
    "Last Update Posted"
]

# Convert date columns to datetime format

for col in date_columns:
    df[col] = pd.to_datetime(df[col], errors="coerce")
```

```python
# Compare dates only when both dates are available
valid_dates = df['Start Date'].notna() & df['Completion Date'].notna()

corrupted_data = df[
    valid_dates &
    (df['Start Date'] > df['Completion Date'])
]

print(f"Records with both dates available: {valid_dates.sum()}")
print(f"Logical date errors: {len(corrupted_data)}")
```

```python
valid_dates = (
    df['Primary Completion Date'].notna() &
    df['Completion Date'].notna()
)

date_errors = df[
    valid_dates &
    (df['Primary Completion Date'] > df['Completion Date'])
]

print(f"Records checked: {valid_dates.sum()}")
print(f"Potential date inconsistencies: {len(date_errors)}")
```

```python
print("Total rows:", len(df))
print("Unique trial IDs:", df['NCT Number'].nunique())
print("Duplicate trial IDs:", df['NCT Number'].duplicated().sum())
```

```python
#drop unnecessary columns
drop_cols = [
    "Rank",
    "Acronym",
    "Study Documents",
    "Other IDs"
]

df.drop(columns=drop_cols, inplace=True)
```
